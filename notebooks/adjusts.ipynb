```python
import pandas as pd
```

```python
df_protestant_canon = pd.read_csv('../resources/canons/protestant_canon.csv', encoding='utf-8-sig')
df_catholic_canon = pd.read_csv('../resources/canons/roman_catholic_canon.csv', encoding='utf-8-sig')
```

```python
df_protestant_canon.tail()
```

```python
df_catholic_canon.tail()
```

```python
df_nvi = pd.read_csv('../resources/bibles/nvi.csv', delimiter=';')
df_ara = pd.read_csv('../resources/bibles/ara.csv', delimiter=';')
```

```python
df_nvi.tail()
```

```python
# replace the book_id with the book_id from the canon. The column sort_index is equal to the book_id in the bible.
df_ara['book_id'] = df_ara['book_id'].apply(lambda x: df_protestant_canon[df_protestant_canon['sort_index'] == x]['book_id'].values[0])
df_nvi['book_id'] = df_nvi['book_id'].apply(lambda x: df_protestant_canon[df_protestant_canon['sort_index'] == x]['book_id'].values[0])
```

```python
# Save the DataFrame to a CSV file
df_ara.to_csv('ara_verses.csv', index=False, encoding='utf-8', sep=';')
df_nvi.to_csv('nvi_verses.csv', index=False, encoding='utf-8', sep=';')
```
